```python
# 📊 Day 7 - Seaborn Basics

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset
data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [55, 60, 65, 70, 75, 80, 85, 90],
    "Score": [45, 50, 58, 65, 70, 78, 85, 92],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR"]
}

df = pd.DataFrame(data)

print("📊 DATASET")
print(df)


# 1️⃣ Scatter Plot
sns.scatterplot(data=df, x="Study_Hours", y="Score")
plt.title("Study Hours vs Score")
plt.show()


# 2️⃣ Bar Plot
sns.barplot(data=df, x="Department", y="Score")
plt.title("Average Score by Department")
plt.show()


# 3️⃣ Box Plot
sns.boxplot(data=df, x="Department", y="Score")
plt.title("Score Distribution by Department")
plt.show()


# 4️⃣ Heatmap
sns.heatmap(df.select_dtypes(include="number").corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()d
```
