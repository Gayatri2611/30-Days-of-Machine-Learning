```python
# ⚙️ Day 12 - Feature Engineering

import pandas as pd

# Create dataset
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Rohan"],
    "Age": [22, 25, 28, 30, 35],
    "Experience": [1, 3, 5, 6, 10],
    "Salary": [30000, 45000, 60000, 70000, 100000]
}

df = pd.DataFrame(data)

print("📊 ORIGINAL DATA")
print(df)


# 1️⃣ Create a new feature

df["Salary_Per_Year_Experience"] = (
    df["Salary"] / df["Experience"]
)

print("\n⚙️ AFTER CREATING NEW FEATURE")
print(df)


# 2️⃣ Create an Age Group feature
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 25, 30, 40],
    labels=["Young", "Mid", "Senior"]
)

print("\n👥 AFTER CREATING AGE GROUP")
print(df)


# 3️⃣ Create an Experience Category
df["Experience_Level"] = pd.cut(
    df["Experience"],
    bins=[0, 2, 5, 10],
    labels=["Beginner", "Intermediate", "Experienced"],
    include_lowest=True
)

print("\n✨ FINAL DATASET WITH NEW FEATURES")
print(df)
```
