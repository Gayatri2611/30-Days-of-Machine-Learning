```python
# 🐼 Day 8 - Pandas Data Manipulation

import pandas as pd

# Create Dataset
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Rohan", "Priya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "HR"],
    "Salary": [50000, 45000, 60000, 55000, None, 45000],
    "Experience": [2, 3, 5, 4, 1, 3]
}

df = pd.DataFrame(data)

print("📊 ORIGINAL DATA")
print(df)


# 1️⃣ Handling Missing Values

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("\n🔹 AFTER HANDLING MISSING VALUES")
print(df)


# 2️⃣ Removing Duplicates

df = df.drop_duplicates()

print("\n🔹 AFTER REMOVING DUPLICATES")
print(df)


# 3️⃣ Filtering Data

print("\n🔹 EMPLOYEES WITH SALARY ABOVE 50000")
print(df[df["Salary"] > 50000])


# 4️⃣ Sorting Data

print("\n🔹 SORTED BY SALARY")
print(df.sort_values(by="Salary", ascending=False))


# 5️⃣ Grouping Data

print("\n🔹 AVERAGE SALARY BY DEPARTMENT")
print(df.groupby("Department")["Salary"].mean())


# 6️⃣ Creating a New Column

df["Salary_Category"] = df["Salary"].apply(
    lambda x: "High" if x >= 55000 else "Low"
)

print("\n🔹 FINAL DATA")
print(df)
```
