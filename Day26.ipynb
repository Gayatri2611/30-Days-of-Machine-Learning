```python
# 🧩 Day 26 - Principal Component Analysis (PCA)

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Create Dataset
data = {
    "Study_Hours": [1, 1.5, 2, 2.5, 3, 4, 5, 6, 6.5, 7, 8, 9],
    "Attendance": [50, 52, 55, 58, 60, 65, 72, 80, 82, 85, 90, 95],
    "Assignments": [2, 2, 3, 3, 4, 4, 5, 6, 6, 7, 8, 9],
    "Test_Score": [40, 42, 45, 48, 52, 58, 65, 72, 75, 80, 88, 92]
}

df = pd.DataFrame(data)

print("📊 ORIGINAL DATASET")
print(df)

# 2. Select Features
X = df[[
    "Study_Hours",
    "Attendance",
    "Assignments",
    "Test_Score"
]]

# 3. Standardize Features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\n📏 DATA AFTER STANDARDIZATION")
print(X_scaled[:5])

# 4. Apply PCA
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

# 5. Create PCA DataFrame
pca_df = pd.DataFrame(
    X_pca,
    columns=["Principal_Component_1", "Principal_Component_2"]
)

print("\n🧩 PCA TRANSFORMED DATA")
print(pca_df)

# 6. Explained Variance
print("\n📈 EXPLAINED VARIANCE RATIO")
print(pca.explained_variance_ratio_)

print("\n📊 TOTAL VARIANCE RETAINED:",
      round(pca.explained_variance_ratio_.sum() * 100, 2), "%")

# 7. Visualize PCA Components
plt.figure(figsize=(8, 6))

plt.scatter(
    pca_df["Principal_Component_1"],
    pca_df["Principal_Component_2"],
    s=100
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA - 2D Visualization")

plt.show()
```
