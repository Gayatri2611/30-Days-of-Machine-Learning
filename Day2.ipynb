```python
# 🤖 Day 2 - Types of Machine Learning

# =====================================================
# 1️⃣ SUPERVISED LEARNING
# =====================================================

from sklearn.linear_model import LogisticRegression

print("🔵 SUPERVISED LEARNING")

# Features: [Study Hours, Attendance]
X = [
    [1, 55],
    [2, 60],
    [3, 65],
    [4, 75],
    [5, 85],
    [6, 90]
]

# Labels: 0 = Fail, 1 = Pass
y = [0, 0, 0, 1, 1, 1]

model = LogisticRegression()
model.fit(X, y)

new_student = [[4.5, 80]]

prediction = model.predict(new_student)

print("New Student:", new_student)
print("Prediction:", "Pass" if prediction[0] == 1 else "Fail")


# =====================================================
# 2️⃣ UNSUPERVISED LEARNING
# =====================================================

from sklearn.cluster import KMeans

print("\n🟢 UNSUPERVISED LEARNING")

# [Annual Income, Spending Score]
customers = [
    [20, 20],
    [25, 25],
    [30, 30],
    [70, 75],
    [75, 80],
    [80, 85]
]

kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(customers)

print("Customer Data:", customers)
print("Customer Groups:", clusters)


# =====================================================
# 3️⃣ SEMI-SUPERVISED LEARNING
# =====================================================

from sklearn.semi_supervised import LabelSpreading

print("\n🟡 SEMI-SUPERVISED LEARNING")

# Some data has labels and some does not
X_semi = [
    [1, 55],
    [2, 60],
    [3, 65],
    [4, 75],
    [5, 85],
    [6, 90]
]

# -1 means the data is unlabelled
y_semi = [0, 0, -1, -1, 1, 1]

semi_model = LabelSpreading()
semi_model.fit(X_semi, y_semi)

print("Original Labels:", y_semi)
print("Learned Labels:", semi_model.transduction_)


# =====================================================
# 4️⃣ REINFORCEMENT LEARNING
# =====================================================

print("\n🔴 REINFORCEMENT LEARNING")

# Agent chooses an action and receives a reward

actions = {
    "Move Left": -1,
    "Move Right": 10
}

print("Actions and Rewards:")

for action, reward in actions.items():
    print(action, "→ Reward:", reward)

best_action = max(actions, key=actions.get)

print("Best Action:", best_action)
```
